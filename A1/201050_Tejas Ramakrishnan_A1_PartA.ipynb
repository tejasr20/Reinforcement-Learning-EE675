# Part A: Chernoff Bound

Suppose $X_1, X_2, \cdots, X_n$ are i.i.d. copies of a $\mathcal{N}(0,\sigma^2)$ r.v. Then for $X = \frac{1}{n} \sum_{i=1}^n X_i$ we know that $$ \mathbb{P} [X \geq \epsilon] \leq \exp\left(\dfrac{-n\epsilon^2}{2\sigma^2}\right)  $$

Write a Python code to run Monte Carlo simulations that verify the inequality. Specifically, for a given  $\epsilon$ and $\sigma$, generate $n$ samples from the zero mean Gaussian distribution $x_1, x_2, \dots, x_n$ and check whether the sample average is more than $\epsilon$. Repeat this experiment 500 times and observe in how many experiments out of those 500 experiments, the sample average is more than $\epsilon$. This will gives us an empirical estimate of ${P}[X\geq \epsilon]$.

Take $\sigma = 0.1,~ \epsilon = 0.05$, and plot the empirical estimate as a function of $n \in \{100, 200, \dots, 1000\}$. In the same plot, include the Chernoff upper bound as a function of $n$. `[10 Marks]`

```python
import numpy as np
import matplotlib.pyplot as plt 
import math 
#importing libraries 
```

```python
eps= 0.05 # epsilon
std= 0.1 # Standard deviation
mu= 0 # mean
n_exp= 500 # number of times we repeat the experiment of sampling n samples from the gaussian distribution 
emp_estimates= [] # List to store emperical estimates 
cbound_estimates= [] # List to store chernoff bounds 
```

```python
l= [i for i in range(1, 51)] # The set of all n 
print(l)
```

```python
for n in l:
    temp= 0
    # print(f"Conducting experiment for n= {n}")
    for i in range(n_exp):
        # Get n samples from the gaussian 
        dist= (np.random.normal(loc= mu, scale= std, size=(n)))
        if((np.mean(dist))>eps):
            # if the sample mean is greater than epsilon 
            temp+=1
    temp/=n_exp
    # Temp now stores the emperical estimate of P(X≥ε)
    print(f"For n={n}, obtained emperical estimate of P(X≥ε)= {temp}")
    emp_estimates.append(temp)
    cbound_estimates.append(np.exp((-n*eps*eps)/(2*std*std)))
    # chernoff upper bound is calculated using trhe formula given 
# print(emp_estimates)
```

```python
print(cbound_estimates)
```

```python
plt.figure()
plt.plot(l, cbound_estimates, color= 'blue', label= 'Chernoff bounds')
plt.plot(l, emp_estimates, color= 'red', label= 'Emperical estimate')
plt.title('Comparision of chernoff upper bound and emperical estimates')
plt.legend()
plt.show()
```
